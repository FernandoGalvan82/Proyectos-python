```
a= 6
b= 7
suma= a + b
print(f"La suma de {a} y {b} es {suma}.")
resta= a - b
print(f"La resta de {a} y {b} es {resta}.")
multiplicacion= a * b
print(f"La multiplicación de {a} y {b} es {multiplicacion}.")
division= a / b
print(f"La división de {a} y {b} es {division}.")
```
