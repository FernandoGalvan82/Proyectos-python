```
data = {'estudiante': ['Ana','Juan','Carla','Pedro','Lucia'],
        'nota': [8,6,9,5,7]}
df = pd.DataFrame (data)
print (df.to_string(index=False))
```

```
promedio = df['nota'].mean()
maxima = df['nota'].max()
minima = df['nota'].min()
print (f'La nota promedio de los estudiantes es: {promedio}')
print (f'La nota maxima es: {maxima}')
print (f'La nota minima es: {minima}')
```
